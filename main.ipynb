```python
from ultralytics import YOLO
import os

# 设置GPU
os.environ['CUDA_VISIBLE_DEVICES'] = '0'

# 加载预训练模型
model = YOLO('/home/aistudio/work/pretain model/yolov8n.pt')  # 推荐使用官方预训练权重

# 训练参数配置
train_args = {
    'data': "/home/aistudio/work/GlobalWheat.yaml",
    'epochs': 50,                  # 小数据集建议50-100轮
    'batch': 8,                     # 小batch更稳定（根据GPU显存调整）
    'imgsz': 640,                   # 输入图像尺寸
    'device': 0,                    # 使用GPU 0
    'lr0': 0.001,                   # 初始学习率（小数据可降至0.0005）
    'lrf': 0.01,                    # 最终学习率 = lr0 * lrf
    'optimizer': 'auto',            # 推荐自动选择（Adam/SGD）
    'weight_decay': 0.0005,         # L2正则化防止过拟合
    'dropout': 0.2,                 # 添加Dropout层（小数据集关键！）
    
    # 数据增强配置（小数据集必须增强）
    'augment': True,
    'mosaic': 0.5,                  # 马赛克增强概率（小数据建议降低）
    'hsv_h': 0.015,                 # 色调增强
    'hsv_s': 0.7,                   # 饱和度增强
    'hsv_v': 0.4,                   # 明度增强
    'flipud': 0.3,                  # 上下翻转概率
    'fliplr': 0.5,                  # 左右翻转概率
    'translate': 0.1,               # 平移增强
    
    # 训练控制
    'patience': 15,                 # 早停等待轮数
    'save': True,                   # 保存最佳模型
    'exist_ok': True,               # 允许覆盖已有训练结果
    'workers': 4,                   # 数据加载线程数
}

# 开始训练
model.train(**train_args)

# 验证模型
metrics = model.val(
    data="/home/aistudio/work/GlobalWheat.yaml",
    batch=16,                       # 验证batch可稍大
    conf=0.25,                      # 置信度阈值
    iou=0.6,                        # IoU阈值
    split='val',                     # 验证集划分
    device=0
)

# 打印关键指标
print(f"mAP@0.5: {metrics.box.map:.4f}")
print(f"Precision: {metrics.box.precision.mean():.4f}")
print(f"Recall: {metrics.box.recall.mean():.4f}")
```

```python
import os
from collections import defaultdict

label_dir = "/home/aistudio/runs/detect/predict4/labels"

# 初始化计数器
count_per_image = defaultdict(int)
count_per_class = defaultdict(int)
total = 0

for txt_file in os.listdir(label_dir):
    if txt_file.endswith(".txt"):
        img_name = txt_file.replace(".txt", ".jpg")  # 假设图片是JPG格式
        with open(os.path.join(label_dir, txt_file), 'r') as f:
            boxes = f.readlines()
        
        # 更新计数器
        count_per_image[img_name] = len(boxes)
        total += len(boxes)
        
        # 统计类别（如果需要）
        for box in boxes:
            class_id = int(box.split()[0])  # 每行第一个值是class_id
            count_per_class[class_id] += 1

# 打印结果
print(f"\n总麦穗穗粒数: {total}")
print("\n每张图片的麦穗穗粒数:")
for img, count in count_per_image.items():
    print(f"{img}: {count}个")
```
